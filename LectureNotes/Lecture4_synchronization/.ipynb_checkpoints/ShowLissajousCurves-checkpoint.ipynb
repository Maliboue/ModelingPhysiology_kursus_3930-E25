# Show Lissajous curves

https://en.wikipedia.org/wiki/Lissajous_curve

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def get_lissajous(n, dt, f1, f2, phi0=0, A=1, B=1):
    t = np.arange(n)*dt
    return t, A*np.sin(2*np.pi*f1*t+phi0), B*np.sin(2*np.pi*f2*t)
```

```python
f1 = 10 # Hz
f2s = np.arange(1,40, 2)
nfigs = f2s.size
```

```python
nrows = 4
ncols = int(np.ceil(nfigs/nrows))
f, axs = plt.subplots(nrows, ncols, figsize=(ncols*3, nrows*3))

for f2, ax in zip(f2s, axs.ravel()):
    t, x, y = get_lissajous(1e4, 0.001, f1, f2, phi0=1)
    ax.plot(x, y, lw=0.1)
    ax.set_title(f"f1/f2 = {f1/f2:.2f}")
    ax.set_aspect(1)
    ax.axis('off')
```
