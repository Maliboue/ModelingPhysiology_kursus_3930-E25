```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

from collections import namedtuple

import plotly.graph_objects as go
```

## Model

Model has many parameters. To keep them in order,
we can use namedtuple or dictionary. Here, we use
namedtuple

```python
# Create a named tuple with its fields (attributes) —- our parameters -- listed in the second input argument
# Values are assigned to each parameter later in the code
params = namedtuple('params', ['C',
                               'g_L', 'g_Na', 'g_K',
                               'E_L', 'E_Na', 'E_K',
                               'V_half_m', 'V_half_n',
                               'k_m', 'k_n',
                               'tau', 'I'])
```

```python
def model(t, state, model_parameters):

    """ Reduced HH model - Eqs. 2.14-2.17

    Parameters
    ----------

        - t : scalar
            Time point at which to evaluate
            dv/dt and dn/dt.

        - state : tuple with 2 elements
            Values of `v` and `n` at which to evaluate
            dv/dt and dn/dt.

        - model_parameters : collections.namedtuple
            Values of model parameters are attributes
            of `model_parameters`.

    Returns
    -------

        - tuple with 2 elements
            dv/dt and dn/dt

    """

    C = model_parameters.C
    g_L = model_parameters.g_L
    g_Na = model_parameters.g_Na
    g_K = model_parameters.g_K
    E_L = model_parameters.E_L
    E_Na = model_parameters.E_Na
    E_K = model_parameters.E_K
    V_half_m = model_parameters.V_half_m
    V_half_n = model_parameters.V_half_n
    k_m = model_parameters.k_m
    k_n = model_parameters.k_n
    tau = model_parameters.tau
    I = model_parameters.I

    v = state[0]
    n = state[1]

    m_inf = 1 / (1 + np.exp((V_half_m - v) / k_m) )
    n_inf = 1 / (1 + np.exp((V_half_n - v) / k_n) )

    I_leak = g_L*(v - E_L)
    I_na = g_Na*m_inf*(v - E_Na)
    I_k = g_K*n*(v - E_K)

    dvdt = (I - I_leak - I_na - I_k) / C
    dndt = (n_inf - n) / tau

    return (dvdt, dndt)
```

## Parameters

```python
# Parameters as in Table 2.3 of the lecture notes

BaseParams = params(
C = 1,
g_L = 8,
E_L = -78,          #for low-threshold I_K for test and Hopf
# E_L = -80,          #for high-threshold I_K used for SN
g_Na = 20,
E_Na = 60,
g_K = 10,
E_K = -90,
V_half_m = -20,
k_m = 15,
tau = 1,           # for test and for Hopf and for SN
V_half_n = -45,    # for low-threshold I_K used for test and for Hopf
# V_half_n = -25    # for high-threshold I_K used for SN
k_n = 5,
I = 40,
)
```

```python
t_span = [0, 200]

dt = 0.01 # sec

t_start = 150  # Olga: We are only interested in the system's dyncamis in the steady state,
               # so we skip the initial part of the solution, where the system is far
               # from the steady state.
               # To see how the system evolves to the steady state, set t_start = 0


t_eval = np.arange(t_start, t_span[1], dt)
```

## Test run (Sec. 1.1.4 in the lecture notes; Fig. 1.1 of the lecture notes)

```python
state_initial = (-60, 0.2)
```

```python
solution = solve_ivp(model, t_span, state_initial, args=[BaseParams,], t_eval=t_eval, max_step=0.01)
```

```python
f, axs = plt.subplots(1,2,figsize=(12, 5))

axs[0].plot(solution.t, solution.y[0])
axs[0].set_xlabel('Time')
axs[0].set_ylabel('V')

axs[1].plot(*solution.y)
axs[1].set_ylabel('n')
axs[1].set_xlabel('V')
```

## Task 1 (first bullet). Hopf biffurcation. Show voltage vs time and phase portrait at different input current, I. Show whole trajectory

```python
Is = np.arange(0, 61, 2)

t_eval = np.arange(0, t_span[1], 0.01)

v_solutions = np.zeros((Is.size, t_eval.size))
n_solutions = np.zeros((Is.size, t_eval.size))

for i, Ii in enumerate(Is):

    p = BaseParams._replace(I=Ii)

    solution = solve_ivp(model, t_span, state_initial,
                      args=[p,], t_eval=t_eval, max_step=0.1)

    v_solutions[i] = solution.y[0]
    n_solutions[i] = solution.y[1]
```

#### PLot in 3D

```python

fig = go.Figure()

t0 = 190 # only show time>t0, i.e., steady state

t_vec = t_eval[t_eval>t0]
vss = v_solutions[:, t_eval>t0]
nss = n_solutions[:, t_eval>t0]

thresh = (0.1, 0.001)

for Ii, vi, ni in zip(Is, vss, nss):
    Ii_vals = np.ones(t_vec.size)*Ii


    if np.std(vi)<thresh[0] and np.std(ni)<thresh[1]:
      vi_point = vi.mean()
      ni_point = ni.mean()
      fig.add_trace(go.Scatter3d(
          x=[Ii], y=[ni_point], z=[vi_point],
          mode='markers',
          marker=dict(size=3, color='red', symbol='circle'),
           showlegend=False
      ))

    else:
      fig.add_trace(go.Scatter3d(
          x=Ii_vals,
          y=ni,
          z=vi,
          mode='lines',
          line=dict(
              color=(0,0,0,1), # (R G B Alpha)
              width=10,
          ), showlegend=False
      ))


# Customize layout
fig.update_layout(
    scene=dict(
        xaxis_title='Input current',
        yaxis_title='n',
        zaxis_title='V',
        xaxis=dict(showline=True, linecolor='black'),
        yaxis=dict(showline=True, linecolor='black'),
        zaxis=dict(showline=True, linecolor='black'),
    ),
    title='V-n phase plane in steady state at different Input current values',
    width=800,
    height=800
)

# Set the camera view
fig.update_layout(scene_camera=dict(
    eye=dict(x=1.35, y=-1.35, z=1.35)  # controls the camera position
))
```

Show full trajectories in separate 2D V-n graphs for each Input current value.

```python
for Ii, vi, ni in zip(Is, v_solutions, n_solutions):

    f, axs = plt.subplots(1,2,figsize=(12, 5))

    plt.suptitle(f'Injected current $I$ = {Ii} mA')
    axs[0].plot(t_eval, vi)
    axs[0].set_xlabel('Time')
    axs[0].set_ylabel('V')
    axs[0].set_ylim(-80, 0)

    axs[1].plot(ni, vi)
    axs[1].set_ylabel('n')
    axs[1].set_xlabel('V')
    # axs[1].set_xlim(-80, 0)
    # axs[1].set_ylim(0, 0.7)
```

## Task 1 (second bullet). Saddle-Node biffurcation. Show voltage vs time and phase portrait at different I. Show whole trajectory

```python
Is = np.arange(0, 31, 2)

t_eval = np.arange(0, t_span[1], 0.01)

v_solutions = np.zeros((Is.size, t_eval.size))
n_solutions = np.zeros((Is.size, t_eval.size))

for i, Ii in enumerate(Is):

    p = BaseParams._replace(I=Ii, E_L=-80, V_half_n=-25)

    solution = solve_ivp(model, t_span, state_initial,
                      args=[p,], t_eval=t_eval, max_step=0.1)

    v_solutions[i] = solution.y[0]
    n_solutions[i] = solution.y[1]
```

```python

fig = go.Figure()

t0 = 185 # only show time>t0, i.e., steady state

t_vec = t_eval[t_eval>t0]
vss = v_solutions[:, t_eval>t0]
nss = n_solutions[:, t_eval>t0]

thresh = (0.1, 0.001)

for Ii, vi, ni in zip(Is, vss, nss):
    Ii_vals = np.ones(t_vec.size)*Ii


    if np.std(vi)<thresh[0] and np.std(ni)<thresh[1]:
      vi_point = vi.mean()
      ni_point = ni.mean()
      fig.add_trace(go.Scatter3d(
          x=[Ii], y=[ni_point], z=[vi_point],
          mode='markers',
          marker=dict(size=3, color='red', symbol='circle'),
           showlegend=False
      ))

    else:
      fig.add_trace(go.Scatter3d(
          x=Ii_vals,
          y=ni,
          z=vi,
          mode='lines',
          line=dict(
              color=(0,0,0,1), # (R G B Alpha)
              width=10,
          ), showlegend=False
      ))


# Customize layout
fig.update_layout(
    scene=dict(
        xaxis_title='Input current',
        yaxis_title='n',
        zaxis_title='V',
        xaxis=dict(showline=True, linecolor='black'),
        yaxis=dict(showline=True, linecolor='black'),
        zaxis=dict(showline=True, linecolor='black'),
    ),
    title='V-n phase plane in steady state at different Input current values',
    width=800,
    height=800
)

# Set the camera view
fig.update_layout(scene_camera=dict(
    eye=dict(x=1.35, y=-1.35, z=1.35)  # controls the camera position
))
```
