```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

```python
def van_der_pol(mu, t_span, y0, t_eval=None, method='RK45'):
    """
    Solves the Van der Pol oscillator ODE numerically.

    Parameters:
    - mu: float, the nonlinearity/stiffness parameter
    - t_span: tuple, (t0, tf) time interval
    - y0: list, initial conditions [x0, y0]
    - t_eval: array-like or None, times to evaluate the solution
    - method: str, solver method ('RK45', 'BDF', etc.)

    Returns:
    - sol: OdeResult from solve_ivp
    """

    def vdp(t, y):
        x, dx = y
        return [dx, mu * (1 - x**2) * dx - x]

    sol = solve_ivp(vdp, t_span, y0, t_eval=t_eval, method=method)
    return sol

# Example usage
mu = 1.2                          # Set to a large value like 1000 for stiff case
t_span = (0, 20)                  # Simulation time range
y0 = [2, 0]                       # Initial conditions: [x(0), x'(0)]
t_eval = np.linspace(*t_span, 500)  # Even time sampling

solution = van_der_pol(mu, t_span, y0, t_eval=t_eval)

# Plotting the solution
plt.plot(solution.t, solution.y[0], label='x(t)')
plt.plot(solution.t, solution.y[1], label="x'(t)")
plt.xlabel('Time')
plt.ylabel('Solution')
plt.title(f'Van der Pol Oscillator (mu={mu})')
plt.legend()
plt.grid(True)
plt.show()
```

Use method='RK45' for non-stiff (mu ≈ 1)

Use method='BDF' for stiff systems (mu ≫ 1)

t_eval ensures evenly spaced output
