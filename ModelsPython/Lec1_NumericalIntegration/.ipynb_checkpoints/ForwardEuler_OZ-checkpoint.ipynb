```python
import numpy as np
import matplotlib.pyplot as plt
```

We want to find a solution, $y(x)$, of an ordinary differential equation:

$$
\frac{\text{d} y(t)}{\text{d}r} = g(t)
$$

For one reason or another, we cannot find the exact solution $y(t)$. Hence, we numerically integrate this equation to find an estimate $\hat{y}(t)$ of the function $y(t)$. 

Euler's integration method is the simplest and the worst numerical integration method, but good for illustration of numerical integration. Euler's method estimates $y$ at discrete time points $t_i$ and requires specification (as any other numerical integrator) of the initial condition $\hat{y}(t_0) = y_0$. All the following values of $\hat{y}(t_i)$, ($i=1,2,\ldots n_{\text{steps}}$) are calculated as follows:

$$
    \hat{y}(t_i) = \hat{y}(t_{i-1}) + (t_i - t_{i-1})~\frac{\text{d} y(t)}{\text{d}r}.
$$

In our case, the time steps between consequitive time points are identical , i.e. $t_i - t_{i-1} = \Delta t$, and

$$
    \hat{y}(t_i) = \hat{y}(t_{i-1}) + \Delta t~\frac{\text{d} y(t)}{\text{d}r}.
$$

---

As examples, we take simple cases, where the exact solutions of the ODE are known. Knowing the exact solutions allows us to test the code we are using for numerical integration. 

Below we numerically integrate two equations: one has exponent of time as a solution, the other has a cosine of time as a solution, and compare the result ($\hat{y}$) to the exact solution $y$.

```python
def euler_integrator(t0, y0, t_step, num_steps, right):
    
    """ Forward Euler integrator
    
    Parameters
    ----------
    
        - t0 : scalar, initial time.
        
        - y0 : scalar, initial value of the function we estimate.
        
        - t_step : scalar, time step.
        
        - num_steps : integer, number of steps.
        
        - right : callable, function returning the first-order derivative of
                    the function we estimate. Receives the function's value
                    as the first argument and corresponding time point as
                    the second value.
                    
    Returns
    -------
    
        - t, 1d-array
            Time values at which the function was estimated.
            
        - y, 1d-array
            Values of the function we estimate.
    
    """
    
    tn = t0 + num_steps * t_step
    
    t = np.arange(t0, tn, t_step)
    
    y = np.zeros(num_steps)
    
    y[0] = y0

    for i in range(1, num_steps):
        
        y[i] = y[i - 1] + t_step * right( y[i - 1], t[i - 1] )
    
    return t, y
```

### Case 1: Exponent

$$
    \frac{\text{d} y(t)}{\text{d}t} = y(t)
$$

```python
def right_side(y, t):
    return y
```

```python
t0 = 0
y0 = 1

t_max = 3

t_theory = np.arange(t0, t_max, 0.01)
y_theory = np.exp(t_theory)
```

```python

step1 = 0.01
plt.plot(*euler_integrator(t0, y0, step1, int(t_max/step1), right=right_side), 'b', label=f'Euler, step={step1}')

step2 = 0.5
plt.plot(*euler_integrator(t0, y0, step2, int(t_max/step2)+1, right=right_side), 'k', label=f'Euler, step={step2}')

plt.plot(t_theory, y_theory, 'r', label='$y(t) = e^{t}$')

plt.legend()
plt.xlabel('t')
plt.ylabel('y')
plt.title('Euler Method vs Exact Solution')
plt.grid(True)
plt.show()
```

### Sine

$$
    \frac{\text{d} y(t)}{\text{d}t} = \text{sin}(t)
$$

```python
def right_side(y, t):
    return np.sin(t)
```

```python
t0 = 0.1
y0 = -1

t_max = 20

t_theory = np.arange(t0, t_max, 0.01)
y_theory = -np.cos(t_theory)
```

```python
step1 = 0.01
plt.plot(*euler_integrator(t0, y0, step1, int(t_max/step1), right=right_side), 'b', label=f'Euler, step={step1}')

step2 = 1
plt.plot(*euler_integrator(t0, y0, step2, int(t_max/step2)+1, right=right_side), 'k', label=f'Euler, step={step2}')

plt.plot(t_theory, y_theory, 'r', label='y(t) = -cos(t)')

plt.legend()
plt.xlabel('t')
plt.ylabel('y')
plt.title('Euler Method vs Exact Solution')
plt.grid(True)
plt.show()
```
