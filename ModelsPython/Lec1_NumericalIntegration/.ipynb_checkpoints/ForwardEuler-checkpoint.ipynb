```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def forward_euler_integrator(func, t_span, y0, tstep):
    """Forward Euler integrator

    Parameters:

    - func : callable
      Right-hand side of the system, i.e., the time derivative of the state y
      at time t. Takes t as the first argument, and state y as the second
      argument.

    - t_span : 2-element sequence
      Interval of integration (t_start, t_end).

    - y0 : scalar
      Initial value of state y. Here, we have a single variable, hence y0 is a
      scalar. In later lectures, we will have multiple variables, which need
      to be integrates; These will be stored in y0 as a sequence.

    - tstep : scalar
      Integration time step.

    Returns:

    - 2-element tupple
      Time vector and y vector.

    Note:

    - parameter names are similar to those used in `scipy.integrate.solve_ivp`
      function, which we will use in later lectures.

    """
    t = np.arange(t_span[0], t_span[1] + tstep/2, tstep)
    y = np.zeros(t.size)
    y[0] = y0

    for i in range(1, t.size):
        y[i] = y[i - 1] + tstep * func(t[i - 1], y[i - 1])

    return t, y
```

Exponent

```python
def model(t, y):
    return -y # dy/dt = -y
```

```python

y0 = 1  # initial condition

tspan = 0, 4.2


for tstep in [0.1, 0.5]:
  plt.plot(*forward_euler_integrator(model, tspan, y0, tstep=tstep), '.-', label=f'Euler, time step={tstep}')

tt = np.linspace(tspan[0], tspan[1], 1000)
plt.plot(tt, np.exp(-tt), 'r', label='Exact')

plt.legend()
plt.xlabel('t')
plt.ylabel('y')
plt.title('Forward Euler Method vs Exact Solution')
plt.grid(True)
plt.show()
```

Sine

```python
def model(t, y):
  return np.sin(t) # dy/dt = sin(t)
```

```python

y0 = -1  # initial condition

tspan = 0, 4*np.pi


plt.figure(figsize=(12,4))

for tstep in [0.1, 0.5]:
  plt.plot(*forward_euler_integrator(model, tspan, y0, tstep=tstep), '.-', label=f'Euler, time step={tstep}')

tt = np.linspace(tspan[0], tspan[1], 1000)
plt.plot(tt, -np.cos(tt), 'r', label='Exact')

plt.legend()
plt.xlabel('t')
plt.ylabel('y')
plt.title('Forward Euler Method vs Exact Solution')
plt.grid(True)
plt.show()
```
