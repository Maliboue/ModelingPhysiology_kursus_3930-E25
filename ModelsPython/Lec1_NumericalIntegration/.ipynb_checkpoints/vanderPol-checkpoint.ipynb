```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

## Model

```python
def vdp_model(t, state, mu=1):
    x, y = state
    return [y, mu * (1 - x**2) * y - x]
```

# Stiff problem

## Model parameters

```python
state_initial = (2, 0)

mu = 1000 
t_span = (0, 3000) # Stiff Van der Pol oscillator
```

## Scipy's recommendation on choosing solver method (https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.solve_ivp.html):

"Explicit Runge-Kutta methods (‘RK23’, ‘RK45’, ‘DOP853’) should be used for non-stiff problems and implicit methods (‘Radau’, ‘BDF’) for stiff problems [9]. Among Runge-Kutta methods, ‘DOP853’ is recommended for solving with high precision (low values of rtol and atol).

If not sure, first try to run ‘RK45’. If it makes unusually many iterations, diverges, or fails, your problem is likely to be stiff and you should use ‘Radau’ or ‘BDF’. ‘LSODA’ can also be a good universal choice, but it might be somewhat less convenient to work with as it wraps old Fortran code."

```python
%time solution_stiff = solve_ivp(vdp_model, t_span, state_initial, method='RK45', args=(mu,))
```

```python
%time solution_stiff = solve_ivp(vdp_model, t_span, state_initial, method='BDF', args=(mu,))
```

## Integration

#### TASK 4: Let the solver determine the time points where the model is evaluated

```python
solution_stiff = solve_ivp(vdp_model, t_span, state_initial, method='BDF', args=(mu,))
```


#### TASK 5: Evaluate the model at specified time points (t_eval)

```python
t_eval = np.linspace(t_span[0], t_span[1], 50)
solution = solve_ivp(vdp_model, t_span, state_initial, t_eval=t_eval, method='BDF', args=(mu,)) # stiff solver
```

### Plot the results

```python
plt.figure(figsize=(10, 5))
plt.title('Van der Pol Oscillator')
plt.xlabel('Time')
plt.ylabel('x (first state variable)')
plt.grid(True)

plt.plot(solution_stiff.t, solution_stiff.y[0], '-o', label='Time points adapted by solve_ivp', alpha=0.5, c='royalblue')
plt.plot(solution.t, solution.y[0], '-o', label='Even time points set by user', c='tomato')

plt.legend()
plt.show()
```
